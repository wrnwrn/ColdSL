# Generate entity descriptions

```python
import pandas as pd
import os


def generate_node_tsv():
    slkg2_raw_data_path = "../data_raw/SLKG2/sldb_complete.csv"
    output_tsv_path = "../data_raw/SLKG2/node_info.tsv"  

    print(f"Reading SLKG2.0: {slkg2_raw_data_path}")
    
    df_kg_raw = pd.read_csv(slkg2_raw_data_path, low_memory=False)

    print("Handling entities fields...")

    
    
    df_kg_entity = df_kg_raw[df_kg_raw["_id"].notna()].copy()

    
    df_kg_entity["_id"] = df_kg_entity["_id"].astype(int)

    
    
    df_kg_entity["_labels"] = df_kg_entity["_labels"].astype(str).str[1:]

    
    if "description" in df_kg_entity.columns:
        df_kg_entity["description"] = df_kg_entity["description"].fillna("")
    else:
        
        print("Warning: 'description' column not found, filling with empty strings.")
        df_kg_entity["description"] = ""

    
    df_kg_entity["name"] = df_kg_entity["name"].fillna("")

    
    
    output_df = df_kg_entity[["_id", "name", "_labels", "description"]]

    
    print(f"Saving to {output_tsv_path} ...")
    output_df.to_csv(output_tsv_path, sep="\t", index=False, encoding="utf-8")  

    print("Done!")


if __name__ == "__main__":
    generate_node_tsv()
```
